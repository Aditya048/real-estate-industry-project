```python
# Now to import my 5 yearly household forecast data by sa2 area

import geopandas as gpd
from pathlib import Path
import pandas as pd
import requests
import openpyxl
```

```python
Current_Path = Path.cwd()
Path_Root = Current_Path.parent.parent

Data_Path = Path_Root/"data"/"Forecasting_Data"
File_Path = Data_Path/"Household_Dwelling_Forecast_2136.xlsx"

Data_Path.mkdir(parents=True, exist_ok=True)

# Now to read in my data :)

Clean_Data = pd.read_excel(File_Path, 
                           sheet_name="Total_Dwellings",
                           header=None
                           )
# Now to write this to my data :)

Csv_File_Path = Data_Path/"Household_Dwelling_Forecast_2136_Cleaneed.csv"
Clean_Data.to_csv(Csv_File_Path)
```

```python
Unclean_Data = pd.read_csv(Csv_File_Path)
```

```python
# Now to remove some meta data and do some minimal cleaning

Dwelling_Data = (
    Unclean_Data
    .iloc[11:, 1:]
    .copy()
)

Dwelling_Data.columns = [
    "GCCSA",
    "SA4_Code",
    "SA3_Code",
    "SA2_Code",
    "Region_Type",
    "SA2_Name",
    "Dwellings_2021",
    "Dwellings_2026",
    "Dwellings_2031",
    "Dwellings_2036",
]


Dwelling_Data = Dwelling_Data[
    Dwelling_Data["Region_Type"] == "SA2"
].copy()

Code_Columns = [
    "SA4_Code",
    "SA3_Code",
    "SA2_Code",
]

Dwelling_Data[Code_Columns] = (
    Dwelling_Data[Code_Columns]
    .astype(str)
)


Dwelling_Columns = [
    "Dwellings_2021",
    "Dwellings_2026",
    "Dwellings_2031",
    "Dwellings_2036",
]

Dwelling_Data[Dwelling_Columns] = (
    Dwelling_Data[Dwelling_Columns]
    .apply(
        pd.to_numeric,
        errors="coerce",
    )
)



Dwelling_Data = (
    Dwelling_Data
    .reset_index(drop=True)
)
```

```python
# Now to write this to a csv

Csv2_File_Path = Data_Path/"Household_Dwelling_Forecast_2136_CleanedV2.csv"
Dwelling_Data.to_csv(Csv2_File_Path)
```
