```python
# Great now that I have done that, play around with this shape file,
# First things first, gotta make the relevant imports
import geopandas as gpd
from pathlib import Path
```

```python
#AusBS_Zones = ABS_Zones, can't name it that cause python sees ABS as a math function
data_path = '../data/geospatial_data/SA2_2021_AUST_GDA2020/data.shp'
AusBS_Zones = gpd.read_file(data_path)
```

```python
# An overveiw of my whole dataframe, seems like:

# SA2_CODE21 provides a unique ID for a localised area
# SA2_NAME21 provides a nume for this localised are

#####################################################################
# CHG_FLAG21 - is a code to describe my best guess, if the data has
# the data has changed or not 3 codes are relevant to VIC, 0, 1, 3
# CHG_LBL21 - 0 = no change, 1 = new, 3 = name change
######################################################################

# SA3_CODE21 - provides a unique ID for a larger area than SA2
# SA3_NAME21 - is the name for SA3_CODE

# SA4_CODE21 - unique code for even larger areas
# SA4_NAME21 - is the name for that

##################################################################
# GCC_CODE21 - unique ID for either rest of vic, greater vic, 
# migratory off shore vic, or no usual location vic
# GCC_NAME21 - is the name as listed above in GCC_CODE21
##################################################################



# STE_CODE21 - what state it is but code for that
# STE_NAME21 - name for relevant code

# AUS_CODE21 - all 'AUS' expect final entry
# AUS_NAME21 - all Australia expect last one, Outside Australia

# AREASQKM21 - Square kilometeres of SA2 zone
# LOCI_URI21 - is a link, yet to go there
# geometry - is geometry yay!!!
AusBS_Zones.head(1)
```

```python
# Now to filter areas only in victoria,

Vic_Filt = (AusBS_Zones["STE_NAME21"] == "Victoria")

AusBS_Vic_Zones = AusBS_Zones[Vic_Filt]
```

```python
AusBS_Vic_Zones.head(1)
```

```python
# Cool may drop the no usual address & Migratory - offshore shipping
AusBS_Vic_Zones["CHG_LBL21"].value_counts(dropna=False)
```

```python
AusBS_Vic_Zones["GCC_NAME21"].unique()
```

```python
##########################################################
# Now to drop some of these rows & coloums  
# that i dont need in this dataframe to save storage
##########################################################

AusBS_Vic_Zones = AusBS_Vic_Zones.drop(columns=[
    "CHG_FLAG21",
    "CHG_LBL21",
    "STE_CODE21",
    "STE_NAME21",
    "AUS_CODE21",
    "AUS_NAME21"
    ]
)

AusBS_Vic_Zones = AusBS_Vic_Zones.dropna(
    subset=["AREASQKM21"]
)
```

```python
# Make sure the file path exists cause of .gitignore
Vic_Zones_Data_Path = Path("../data/area_zones_data")
Vic_Zones_Data_Path.mkdir(parents=True, exist_ok=True)

AusBS_Vic_Zones.to_file(Vic_Zones_Data_Path/"Vic_Area_Zones.shp")
```
