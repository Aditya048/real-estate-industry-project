```python
with sqlite3.connect(DB_PATH) as conn:
    df = pd.read_sql("SELECT * FROM property_details", conn, parse_dates=["added_at"])

# restore the types SQLite flattens
for col in ["furnished", "partially_furnished", "pets_allowed",
            "long_term_lease", "is_build_to_rent"]:
    df[col] = df[col].astype("boolean")

df
```
